````python
import pandas as pd
from sqlalchemy import create_engin
import os
from dotenv import load_dotenv
load_dotenv()

db_config = {'user': os.getenv("user"), # имя пользователя
             'pwd': os.getenv("pwd"), # пароль
             'host': os.getenv("host"),
             'port': os.getenv("port"), # порт подключения
             'db': os.getenv("db") # название базы данных
             }
connection_string = 'postgresql://{}:{}@{}:{}/{}'.format(
    db_config['user'],
    db_config['pwd'],
    db_config['host'],
    db_config['port'],
    db_config['db'],
)
engine = create_engine(connection_string)
query = '''
SELECT
    p.user_id,
    p.device_type_canonical,
    p.order_id,
    p.created_dt_msk AS order_dt,
    p.created_ts_msk AS order_ts,
    p.currency_code,
    p.revenue,
    p.tickets_count,

    p.created_dt_msk - LAG(p.created_dt_msk) OVER (
        PARTITION BY p.user_id
        ORDER BY p.created_ts_msk
    ) AS days_since_prev,

    p.event_id,
    p.service_name,
    e.event_type_main,
    r.region_name,
    c.city_name

FROM afisha.purchases AS p

LEFT JOIN afisha.events AS e
    ON p.event_id = e.event_id

LEFT JOIN afisha.city AS c
    ON e.city_id = c.city_id

LEFT JOIN afisha.regions AS r
    ON c.region_id = r.region_id
'''

df = pd.read_sql_query(query, con=engine)
df
````

**Задача 2.1:** Данные о выручке сервиса представлены в российских рублях и казахстанских тенге. Приведите выручку к единой валюте — российскому рублю.

Для этого используйте датасет с информацией о курсе казахстанского тенге по отношению к российскому рублю за 2024 год — `final_tickets_tenge_df.csv`. Его можно загрузить по пути `https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv')`

Значения в рублях представлено для 100 тенге.

Результаты преобразования сохраните в новый столбец `revenue_rub`.

---


````python
# Загружаем данные о курсе тенге
tenge_df = pd.read_csv(
    'https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv'
)

# Приводим даты к единому формату
tenge_df['data'] = pd.to_datetime(tenge_df['data'])
df['order_dt'] = pd.to_datetime(df['order_dt'])

# Объединяем данные о покупках с курсом тенге по дате
df = df.merge(
    tenge_df[['data', 'curs']],
    left_on='order_dt',
    right_on='data',
    how='left'
)

# Создаём новый столбец
df['revenue_rub'] = df['revenue']

# Пересчитываем выручку для заказов в тенге
mask = df['currency_code'] == 'kzt'

df.loc[mask, 'revenue_rub'] = (
    df.loc[mask, 'revenue'] * df.loc[mask, 'curs'] / 100
)

# Удаляем вспомогательные столбцы
df = df.drop(columns=['data', 'curs'])

df.head()
````

### 3. Создание профиля пользователя

В будущем отдел маркетинга планирует создать модель для прогнозирования возврата пользователей. Поэтому сейчас они просят вас построить агрегированные признаки, описывающие поведение и профиль каждого пользователя.

---

**Задача 3.1.** Постройте профиль пользователя — для каждого пользователя найдите:

- дату первого и последнего заказа;
- устройство, с которого был сделан первый заказ;
- регион, в котором был сделан первый заказ;
- билетного партнёра, к которому обращались при первом заказе;
- жанр первого посещённого мероприятия (используйте поле `event_type_main`);
- общее количество заказов;
- средняя выручка с одного заказа в рублях;
- среднее количество билетов в заказе;
- среднее время между заказами.

После этого добавьте два бинарных признака:

- `is_two` — совершил ли пользователь 2 и более заказа;
- `is_five` — совершил ли пользователь 5 и более заказов.

**Рекомендация:** перед тем как строить профиль, отсортируйте данные по времени совершения заказа.

---


````python
# Шаг 1. Сортируем данные по пользователю и времени заказа.
# Это необходимо, чтобы first() выбирал данные именно первого заказа.
df = df.sort_values(['user_id', 'order_ts'])

# Шаг 2. Создаём числовой столбец с количеством дней между заказами.
# Для первого заказа пользователя значение останется NaN,
# так как предыдущего заказа нет.
df['days_between_orders'] = df['days_since_prev'].dt.days

# Шаг 3. Формируем профиль пользователя
user_profile = (
    df.groupby('user_id')
      .agg(
          first_order_dt=('order_dt', 'min'),                 
          last_order_dt=('order_dt', 'max'),                 
          first_device=('device_type_canonical', 'first'),   
          first_region=('region_name', 'first'),             
          first_service=('service_name', 'first'),           
          first_event_type=('event_type_main', 'first'),     
          orders_count=('order_id', 'nunique'),              
          avg_revenue_rub=('revenue_rub', 'mean'),            
          avg_tickets=('tickets_count', 'mean'),             
          avg_days_between_orders=('days_between_orders', 'mean')
      )
      .reset_index()
)

# Шаг 4. Округляем средние показатели
user_profile['avg_revenue_rub'] = user_profile['avg_revenue_rub'].round(2)
user_profile['avg_tickets'] = user_profile['avg_tickets'].round(2)
user_profile['avg_days_between_orders'] = (
    user_profile['avg_days_between_orders'].round(2)
)

# Шаг 5. Проверяем результат
user_profile.head()
````

````python
# Создаём бинарный признак is_two:
# 1 — пользователь совершил 2 и более заказа,
# 0 — пользователь совершил только 1 заказ
user_profile['is_two'] = (user_profile['orders_count'] >= 2).astype(int)

# Создаём бинарный признак is_five:
# 1 — пользователь совершил 5 и более заказов,
# 0 — пользователь совершил менее 5 заказов
user_profile['is_five'] = (user_profile['orders_count'] >= 5).astype(int)

# Проверяем результат
user_profile[['user_id', 'orders_count', 'is_two', 'is_five']].head(10)
````

При интерпретации результатов учитывайте размер сегментов: если в сегменте мало пользователей (например, десятки), то доли могут быть нестабильными и недостоверными, то есть показывать широкую вариацию значений.

````python
# Определяем сегмент каждого пользователя —
# тип его первого события
user_segments = (
    df.sort_values('order_ts')
      .groupby('user_id')
      .agg(first_event_type=('event_type_main', 'first'))
      .reset_index()
)

# Считаем количество уникальных заказов каждого пользователя
user_orders = (
    df.groupby('user_id')
      .agg(orders_count=('order_id', 'nunique'))
      .reset_index()
)

# Объединяем сегмент и количество заказов
user_data = user_segments.merge(
    user_orders,
    on='user_id',
    how='left'
)

# Для каждого сегмента считаем:
# 1. общее количество пользователей
# 2. количество пользователей с 2+ заказами
# 3. их долю
repeat_share = (
    user_data
    .groupby('first_event_type')
    .agg(
        users_count=('user_id', 'count'),
        repeat_users=('orders_count', lambda x: (x >= 2).sum())
    )
    .reset_index()
)

repeat_share['repeat_share'] = (
    repeat_share['repeat_users'] /
    repeat_share['users_count'] * 100
).round(2)

repeat_share.sort_values('repeat_share', ascending=False)
````

````python
# 10 сегментов с наибольшим количеством пользователей
top_10 = (
    repeat_share
    .nlargest(10, 'users_count')
    .sort_values('repeat_share')
)

# Визуализация
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.barh(
    top_10['first_event_type'],
    top_10['repeat_share']
)

plt.xlabel('Доля пользователей с 2+ заказами, %')
plt.ylabel('Сегмент')
plt.title('Доля пользователей с повторными заказами по сегментам')

# Подписываем значения на столбцах
for i, value in enumerate(top_10['repeat_share']):
    plt.text(
        value + 0.3,
        i,
        f'{value:.1f}%',
        va='center'
    )

plt.tight_layout()
plt.show()
````

````python

# Приводим дату заказа к формату datetime
df['order_dt'] = pd.to_datetime(df['order_dt'])

# Находим дату первого заказа каждого пользователя
first_order_date = (
    df.groupby('user_id')['order_dt']
    .min()
    .reset_index(name='first_order_date')
)

# Выделяем день недели
first_order_date['first_order_weekday'] = (
    first_order_date['first_order_date']
    .dt.day_name()
)

# Добавляем информацию о повторных заказах
first_order_date = first_order_date.merge(
    user_region_data[['user_id', 'is_returned']],
    on='user_id',
    how='left'
)

# Статистика по дням недели
weekday_stats = (
    first_order_date
    .groupby('first_order_weekday')
    .agg(
        users_count=('user_id', 'count'),
        repeat_users=('is_returned', 'sum')
    )
    .reset_index()
)

# Доля пользователей с повторными заказами
weekday_stats['repeat_share'] = (
    weekday_stats['repeat_users'] /
    weekday_stats['users_count'] * 100
).round(2)

# Правильный порядок дней недели
weekday_order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

weekday_stats['first_order_weekday'] = pd.Categorical(
    weekday_stats['first_order_weekday'],
    categories=weekday_order,
    ordered=True
)

weekday_stats = weekday_stats.sort_values('first_order_weekday')

weekday_stats
````

````python
# Переводим интервал между заказами в количество дней
df['days_since_prev_num'] = (
    df['days_since_prev'].dt.total_seconds() / (24 * 60 * 60)
)
# Среднее время между заказами для каждого пользователя
avg_days_between_orders = (
    df.groupby('user_id')['days_since_prev_num']
    .mean()
    .reset_index(name='avg_days_between_orders')
)

# Добавляем количество заказов пользователя
user_orders_data = avg_days_between_orders.merge(
    user_region_data[['user_id', 'orders_count']],
    on='user_id',
    how='left'
)

# Оставляем пользователей с 2 и более заказами
user_orders_data = user_orders_data[
    user_orders_data['orders_count'] >= 2
]

# Разделяем пользователей на группы
user_orders_data['order_group'] = np.where(
    user_orders_data['orders_count'].between(2, 4),
    '2–4 заказа',
    '5 и более заказов'
)

# Считаем среднее время между заказами для каждой группы
time_comparison = (
    user_orders_data
    .groupby('order_group')
    .agg(
        avg_days_between_orders=('avg_days_between_orders', 'mean')
    )
    .reset_index()
)

time_comparison['avg_days_between_orders'] = (
    time_comparison['avg_days_between_orders']
    .round(2)
)

time_comparison
````
